duplicate

```python
import pandas as pd

data = {
    'Name':['John','Mary', 'John'],
    'Sales':[1000,1500,1000]
}

df = pd.DataFrame(data)

print(df.duplicated())
```

dupllicate remove

```python
import pandas as pd

data = {
    'Name':['John','Mary', 'John'],
    'Sales':[1000,1500,1000]
}

df = pd.DataFrame(data)

df = df.drop_duplicates()

print(df)
```

rename

```python
import pandas as pd

data = {
    'emp_nm':['john','Mary'],
    'emp_sal':[25000,30000]
}
df = pd.DataFrame(data)

df.rename(columns={

    'em_nm':'employee Name',
    'em_sal':'employee salary'
}, inplace=True)

print(df)
```

data analysis

```python
import pandas as pd

data = {
    "Region": ["south", "North", "East", "west",],
    "Product": ["Laptop", "Mobile", "Laptop", "Tablet"],
    "Sales": [50000, 30000, 45000, 25000]

}

df = pd.DataFrame(data)

print(df)
```

Filtring

```python
south_sales = df[df["Region"] == "south"]

print(south_sales)
```
