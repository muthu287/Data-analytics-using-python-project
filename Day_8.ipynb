```python
import matplotlib.pyplot as plt

months = ["Jan", "Feb", "Mar", "Apr", "May"]

sales =  [10000, 12000, 15000, 13000, 18000]

plt.plot(months,sales)

plt.title("Months")

plt.xlabel("sales Amount")

plt.show()
```

```python
import matplotlib.pyplot as plt

students = ["A","B","c","D"]

marks = [80,100,40,90]

plt.bar(students,marks)

plt.title("students Marks")

plt.xlabel("students")

plt.ylabel("Marks")

plt.show()
```

```python
import matplotlib.pyplot as plt

months = ["Jan", "Feb", "Mar", "Apr", "May"]

sales = [10000, 12000,15000,13000,18000]

plt.plot(months, sales,
         color="green",
         marker="o",
         linestyle="--")

plt.title("Sales Trend")

plt.xlabel("Month")

plt.ylabel("Sales")

plt.show()
```

```python
import matplotlib.pyplot as plt

months = ["Jan","Feb","Mar","Apr"]

sales = [1000,12000,15000,20000]

plt.figure(figsize=(12,5))

# Line chart
plt.subplot(1,2,1)

plt.plot(months,sales,marker="o")

plt.title("sales Trend")

plt.xlabel("Months")

plt.ylabel("sales")

# Bar chart
plt.subplot(1,2,2)

plt.bar(months,sales,color="orange")

plt.title("sales comparision")
```
